**Q1.** What is a Common Table Expression (CTE), and how does it improve SQL query readability?

**Ans-** **COMMON TABLE EXPRESSION (CTE) :** A Common Table Expression (CTE) is a temporary named result set that is defined using the WITH clause and exists only for the duration of a single SQL statement. It allows you to break a complex query into smaller, more manageable parts, making the query easier to read, write, and maintain.

A CTE is not permanently stored in the database. It behaves like a temporary table that can be referenced multiple times within the same query.

**SYNTAX:**
  
    WITH cte_name AS (
    SELECT column1, column2
    FROM table_name
    WHERE condition
    )
    SELECT * FROM cte_name;

### **How does a CTE improve SQL query readability?**

**1.** **Improves Readability**

- Breaks large and complex queries into smaller, logical sections.
- Makes the query easier to understand.

**2.** **Reduces Code Repetition**

- A CTE can be referenced multiple times within the same query, avoiding repeated subqueries.

**3.** **Simplifies Complex Queries**

-  Makes queries involving multiple joins, aggregations, or calculations more organized.

**4.** **Easier Maintenance**

- Changes only need to be made in one place, making the query easier to update and debug.

**5.** **Supports Recursive Queries**

- CTEs can be used to retrieve hierarchical data, such as employee-manager relationships or category trees.

### **ADVANTAGES OF CTE**

- Enhances query readability.
- Improves code organization.
- Eliminates repeated subqueries.
- Simplifies debugging and maintenance.
- Supports recursive operations.
- Makes complex SQL queries easier to understand.










**Q2.** Why are some views updatable while others are read-only? Explain with an example.

**Ans-** A view is a virtual table that is created from the result of a SQL query. Whether a view is updatable or read-only depends on how it is defined.

**UPDATABLE VIEW**

A view is updatable when it is created from a single base table and does not contain complex SQL operations such as JOIN, GROUP BY, DISTINCT, aggregate functions (SUM, AVG, COUNT), UNION, or calculated columns. In this case, the database can clearly identify which row in the base table should be modified.

**Q3.** What advantages do stored procedures offer compared to writing raw SQL queries repeatedly?

**Ans-** A stored procedure is a precompiled collection of one or more SQL statements stored in the database. It is created once and can be executed multiple times whenever required. Compared to writing raw SQL queries repeatedly, stored procedures provide several advantages in terms of performance, security, maintainability, and reusability.

### **Advantages of Stored Procedures**

1. **Improved Performance**
- Stored procedures are precompiled and their execution plans are cached by the database.
- This reduces query parsing and compilation time, resulting in faster execution.
2. **Code Reusability**
- A stored procedure is written once and can be called multiple times by different users or applications.
- This eliminates duplicate SQL code and improves productivity.
3. **Enhanced Security**
- Users can be granted permission to execute a stored procedure without providing direct access to the underlying tables.
- This protects sensitive data and reduces the risk of unauthorized access and SQL injection.
4. **Easy Maintenance**
- Business logic is centralized within the stored procedure.
- Any modifications are made only once, and the updated logic is automatically available to all applications that use it.
5. **Reduced Network Traffic**
- Instead of sending multiple SQL statements over the network, the application sends only a procedure call.
- This minimizes communication between the application and the database server.
6. **Consistency**
- Since the same stored procedure is used every time, it ensures consistent implementation of business rules and reduces coding errors.
7. **Parameter Support**
- Stored procedures accept input parameters and can return output values.
- This makes them flexible and reusable for different input values without changing the SQL code.

**Q4.** What is the purpose of triggers in a database? Mention one use case where a trigger is essential.

**Ans-** A trigger is a special type of stored program in a database that automatically executes (fires) in response to a specific event such as INSERT, UPDATE, or DELETE on a table. Unlike stored procedures, triggers are not called explicitly; they are invoked automatically whenever the specified event occurs.

The primary purpose of a trigger is to automate database operations, enforce business rules, maintain data integrity, and ensure consistency without requiring manual intervention or additional application code.

### **Purpose of Triggers**

1. **Automate database operations** by executing predefined actions automatically after or before data changes.
2. **Maintain data integrity** by enforcing business rules and preventing invalid data modifications.
3. **Create audit logs** to record changes made to important data for tracking and security purposes.
4. **Validate data** before it is inserted or updated in the database.
5. **Keep related tables synchronized** automatically whenever data changes occur.

### **Use Case: Maintaining an Audit Log**

A common and essential use case for a trigger is recording employee salary changes in an audit table. Whenever an employee's salary is updated, the trigger automatically stores the old salary, new salary, and the date of the change.

**Q5.** Explain the need for data modelling and normalization when designing a database.

**Ans-** When designing a database in SQL, it is important to organize data in a way that ensures accuracy, consistency, efficiency, and easy maintenance. Two fundamental concepts used to achieve this are Data Modelling and Normalization.

### **1. Data Modelling**

Data modelling is the process of designing the logical structure of a database before creating it in SQL. It identifies the entities (tables), attributes (columns), relationships, and constraints required to represent real-world data.

**Need for Data Modelling**

- Provides a blueprint for designing the SQL database.
- Identifies tables, columns, primary keys, and foreign keys.
- Defines relationships between tables (One-to-One, - One-to-Many, Many-to-Many).
- Reduces design errors and data duplication.
- Makes the database scalable, maintainable, and easy to understand.
- Ensures the database meets business requirements.

**EXAMPLE**

### **2. Normalization**

Normalization is the process of organizing data into multiple related tables to eliminate data redundancy and maintain data integrity. It ensures that each piece of information is stored only once.

**Need for Normalization**

- Eliminates duplicate data.
- Maintains data consistency and integrity.
- Prevents Insert, Update, and Delete anomalies.
- Reduces storage space.
- Simplifies database maintenance.
- Improves query efficiency and database performance.

| **Data Modelling**                              | **Normalization**                       |
| ----------------------------------------------- | --------------------------------------- |
